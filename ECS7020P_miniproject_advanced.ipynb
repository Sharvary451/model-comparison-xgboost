```python
import pandas as pd
from sklearn.model_selection import train_test_split
#MACHINE LEARNING PIPELINE
#Input
# In the code below,we load the dataset into a dataframe named 'dfingredients' to be used later in the code
dfingredients = pd.read_csv('..\\MLEnd\\yummy\MLEndYD_image_attributes_benchmark.csv')
#Data preprocessing
# In the Exccel provided,there is a column named 'Ingredients' that I have referred to. There were some values that were NaN
#Hence,I had to drop those values so that they wouldn't cuase an issue later on.
#In the code below, the Dataframe created above is used to extract 'Ingredients' and store it in a variable 'ingredients_list'
ingredients_list = dfingredients['Ingredients'].dropna()  # Drop rows which have NaN values
#TRAIN-TEST SPLIT
#In teh code below,we split the data into training data and testing data.The test_size is the parameter passed to specify 
#that 20% of the data will be used for testing while 80% of the data will be used for training. random_state is used to 
#specify a value that is to be used for randomness. This is important so that same split occours each time model is run.
train_data, test_data = train_test_split(ingredients_list, test_size=0.2, random_state=40)
#TRANSFORMATION
#Feature Extraction
# In the code below,a function 'cal_freq' is defined which is used to create a dictionary which will store the number 
#of times each ingredient occurs. An empty dictionary called 'ingredient_frequency' is initialized which will store the
#results. The for loop iterates through each ingredient present in the list. the split and strip function are used to 
#create commas and remove any whitespaces.The second 'For loop' checks if the ingredient has already been accounted for
#If yes.the value is incremented by 1 and the next ingredient is taken into consideration,if not,that ingredient is added
#to the dictionary.
def cal_freq(data):
    ingredient_frequency = {}
    for ingredients in data:
        ingredient_list = ingredients.split(',') 
        for ingredient in ingredient_list:
            ingredient = ingredient.strip()  
            if ingredient in ingredient_frequency:
                ingredient_frequency[ingredient] += 1
            else:
                ingredient_frequency[ingredient] = 1
    return ingredient_frequency

# In the code below,we use the function defined above to calculate the frequencies i.e. number of times the 
#ingredients appear in both test and training data. This is used to calculate which ingredient is the most popular in
#both datasets
train_frequency = cal_freq(train_data)
test_frequency = cal_freq(test_data)

# In the code below,a variable 'mostfreq' is initialized which sets the value as 30. This means that the top 30 most 
#popular ingredients will be considered and printed. The line below that depicts the sorting of ingredients. It creates 
#tuple in which ingredients and its frequency are stored. This is for training data
mostfreq= 30
print(f"Top {mostfreq} Popular Ingredients in Training Set:")
sorted_train_ingredients = sorted(train_frequency.items(), key=lambda x: x[1], reverse=True)
#In the below for loop,it iterates the all ingredients and the ingrdient which is stored in mostfreq is the ingredient
#which appears the most number of times,hence,the most popular ingredient
#RESULTS
for ingredient, frequency in sorted_train_ingredients[:mostfreq]:
    print(f"{ingredient}: {frequency} occurrences")

print("\n")

# In the code below,a variable 'mostfreq' is initialized which sets the value as 30. This means that the top 30 most 
#popular ingredients will be considered and printed. The line below that depicts the sorting of ingredients. It creates 
#tuple in which ingredients and its frequency are stored. This is for testing data
print(f"Top {mostfreq} Popular Ingredients in Testing Set:")
sorted_test_ingredients = sorted(test_frequency.items(), key=lambda x: x[1], reverse=True)
#In the below for loop,it iterates the all ingredients and the ingrdient which is stored in mostfreq is the ingredient
#which appears the most number of times,hence,the most popular ingredient
#RESULTS
for ingredient, frequency in sorted_test_ingredients[:mostfreq]:
    print(f"{ingredient}: {frequency} occurrences")
```
